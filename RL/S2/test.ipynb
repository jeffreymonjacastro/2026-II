```python
def compute_return(trajectory, gamma):
    """Calcula el retorno descontado desde t=0."""
    G = 0
    print(G)
    for _, _, r in reversed(trajectory):
        G = r + gamma * G
        print(G)
    return G
```

```python
traj = [(0, 0, 1), (0, 1, 2), (1, 0, 3)]
# print(list(reversed(traj)))

compute_return(traj, 0.9)
```
